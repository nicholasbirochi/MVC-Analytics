```python
# Funções para corrigir os tipos das colunas, as colnas que vão ser usadas e a conversão:
def converter_colunas_mistas_para_str(df):
    for col in df.columns:
        if df[col].apply(type).nunique() > 1:
            df[col] = df[col].astype(str)
    return df


def ler_sem_colunas(path, colunas_excluir, **kwargs):
    colunas = pd.read_csv(path, nrows=0, **kwargs).columns.tolist()

    if all(isinstance(c, int) for c in colunas_excluir):
        colunas_ler = [col for i, col in enumerate(colunas) if i not in colunas_excluir]
    else:
        colunas_ler = [col for col in colunas if col not in colunas_excluir]

    return pd.read_csv(path, usecols=colunas_ler, **kwargs)


def corrigir_csv(path, colunas_excluir=None, renomear_dict=None, nova_ordem=None, encoding_origem="utf-8", encoding_saida="utf-8", newpath=None):
    """
    Lê um CSV, remove colunas, renomeia e reordena, converte tipos mistos e salva o arquivo corrigido.

    Parâmetros:
      - path (str): caminho do CSV
      - colunas_excluir (list): nomes ou índices a remover
      - renomear_dict (dict): mapeamento de nomes antigos → novos
      - nova_ordem (list): lista com a ordem desejada das colunas
      - encoding_origem (str): encoding de entrada
      - encoding_saida (str): encoding de saída
      - newpath (str): novo caminho do CSV
    """

    # === Ler CSV removendo colunas indesejadas ===
    df = ler_sem_colunas(path, colunas_excluir or [], encoding=encoding_origem)

    # === Renomear colunas específicas ===
    if renomear_dict:
        df.rename(columns=renomear_dict, inplace=True)

    # === Reordenar colunas ===
    if nova_ordem:
        # Mantém apenas colunas que realmente existem no DataFrame
        colunas_existentes = [c for c in nova_ordem if c in df.columns]
        outras = [c for c in df.columns if c not in colunas_existentes]
        df = df[colunas_existentes + outras]

    # === Corrigir tipos mistos ===
    df = converter_colunas_mistas_para_str(df)

    # === Salvar CSV final ===
    df.to_csv(newpath, index=False, encoding=encoding_saida)

    print(f"✅ Arquivo corrigido e salvo em: {newpath}")
```
