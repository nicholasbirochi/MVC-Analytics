```python
# normaliza os campos
df_pessoas["CHAPA"] = df_pessoas["CHAPA"].astype(str).str.strip()
df_ronda["cadastro"] = df_ronda["cadastro"].astype(str).str.strip()

# pessoas que estão em df_pessoas mas não aparecem em df_ronda
df_nao_aparecem = df_pessoas[~df_pessoas["CHAPA"].isin(df_ronda["cadastro"])].copy()

print('Pessoas que não estão na base de registros do Ronda:')
print(df_nao_aparecem)
```

```python
# lista que você quer checar
chapas = [
    "9771252",
    "9771239",
    "9771141",
    "9771014",
    "9798139"
]

# normalizar
df_pessoas["CHAPA"] = df_pessoas["CHAPA"].astype(str).str.strip()
df_ronda["cadastro"] = df_ronda["cadastro"].astype(str).str.strip()

# dataframe com as chapas consultadas
df_consulta = pd.DataFrame({"CHAPA": chapas})

# verificar presença em cada base
df_consulta["em_df_pessoas"] = df_consulta["CHAPA"].isin(df_pessoas["CHAPA"])
df_consulta["em_df_ronda"] = df_consulta["CHAPA"].isin(df_ronda["cadastro"])

# classificação
df_consulta["onde_esta"] = "nenhum"
df_consulta.loc[
    df_consulta["em_df_pessoas"] & ~df_consulta["em_df_ronda"],
    "onde_esta"
] = "só df_pessoas"

df_consulta.loc[
    ~df_consulta["em_df_pessoas"] & df_consulta["em_df_ronda"],
    "onde_esta"
] = "só df_ronda"

df_consulta.loc[
    df_consulta["em_df_pessoas"] & df_consulta["em_df_ronda"],
    "onde_esta"
] = "em ambos"

print('Vendo se as pessoas específicas solicitadas pelo Fernando estão na base do Ronda:')
print(df_consulta)
```
